```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style='whitegrid')

date_columns = ['last_scraped', 'calendar_last_scraped', 'first_review', 'last_review']
listings = pd.read_csv('data/listings_clean.csv', parse_dates=date_columns)

print(listings.shape)
```

## Lab 8
### Ex 1

```python
price_night = listings['price']
no_price = listings['price'].isna()
long_min = listings['minimum_nights'] > 7
price_long = listings.loc[long_min, 'price']
vlong_min = listings['minimum_nights'] > 28
price_vlong = listings.loc[vlong_min, 'price']
percent_price = (price_night.count() / listings['id'].count()).round(2) * 100
cases = [('base', price_night), 
         ('long', price_long), 
         ('very long', price_vlong)]

print('listings with price:', price_night.count(),
      'percent of all:', percent_price)

for desc,case in cases:
    print(f'{desc} mean:', case.mean().round(1),
          f'{desc} median:', case.median())
```

- Claim 1: "A night in Stockholm costs 1,791."
- Hold for very non-stated specific case.
- New sentence: "During end of September 2026 a short stay in Stockholm had the median price per night of 1791SEK. According to the available 2957 listings on AirBnB which is was about 65% of the total."
- Could add: "For longer stays of a week this median drops to half and for month long bookings the per night price is about 40% that of short period bookings."
### Ex 2

```python
rated = listings['has_reviews']
score_5 = listings['review_scores_rating'] == 5.0
reviews_20 = listings['number_of_reviews'] >= 20

print('listings with review', listings[rated]['id'].count(),
     'percent of listings', (listings[rated]['id'].count() / listings['id'].count()).round(3) * 100)
print('listings with 5.0 rating', listings[score_5]['id'].count(),
     'percent of listings', (listings[score_5]['id'].count() / listings['id'].count()).round(3) * 100)
print('listings with 5.0 rating and 20 reviews', listings[score_5 & reviews_20]['id'].count(),
     'percent of listings', (listings[score_5 & reviews_20]['id'].count() / listings['id'].count()).round(3) * 100)
print(listings[score_5]['number_of_reviews'].median())

fig, ax = plt.subplots()

sns.histplot(data=listings[score_5], x='number_of_reviews', bins=200, ax=ax)
ax.bar_label(ax.containers[0])
ax.set_xlim(0, 20)

ax.set_title('Listings with 5.0 score rating.')
ax.set_xlabel('Number of reviews')
ax.set_ylabel('Number of listings')

plt.tight_layout()

plt.show()
```

- Claim 2: "The listings with a perfect 5.00 rating are the best in Stockholm."
- Verdict: Does Not Hold

- Sentence: "Only about 86% of listings have a review. Out of the reviewed listings 32.4% have a perfect score. And out of these perfect score listings less than 3% have 20 or more reviews. A perfect score of 5.0 is more a sign of less reviews than being one of the best listings."

### Ex 4

```python
print(listings[['host_id', 'host_listings_count']].sort_values('host_listings_count', ascending=False).head(10))
# many duplicate ids. because they are per listing.
no_dup = listings.drop_duplicates('host_id')
print(no_dup[['host_id', 'host_listings_count']].sort_values('host_listings_count', ascending=False).head(10))
# after cleaning duplicates not many multi-listing hosts left
print(no_dup.loc[no_dup['host_listings_count'] > 2, 'host_id'].count())
n_list = listings['id'].count()
exact1 = listings.loc[listings['host_listings_count'] == 1, 'id'].count()
exact2 = listings.loc[listings['host_listings_count'] == 2, 'id'].count()
above2 = listings.loc[listings['host_listings_count'] > 2, 'id'].count()
above5 = listings.loc[listings['host_listings_count'] > 5, 'id'].count()
print('host with 1 listing', exact1, (exact1 / n_list).round(3) * 100, '%')
print('host with 2 listing', exact2, (exact2 / n_list).round(3) * 100, '%')
print('host with >2 listing', above2, (above2 / n_list).round(3) * 100, '%')
print('host with >5 listing', above5, (above5 / n_list).round(5) * 100, '%')

fig, ax = plt.subplots()

sns.histplot(data=no_dup, x='host_listings_count', bins=50, ax=ax)
ax.set_xlim(0, 100)
ax.set_ylim(0, 30)
ax.bar_label(ax.containers[0])

ax.set_title('Most hosts have less than 20 listings.\n'+
            f'almost all hosts have 1-2 listings (1:{exact1}, 2:{exact2})\n'+
            'chart y-axis cut off not clearly representing the lower amounts')
ax.set_xlabel('number of listings')
ax.set_ylabel('hosts')
```

- Claim 4: "Most hosts in Stockholm are professional companies with many
listings."
- Verdict: Does not hold
- Sentence: "Overwhelming amount of listings are by hosts with 1-2 listings, 90%+. There are outlier hosts with many listings, clearly companies. But these represent a small portion of both total listings and total hosts."
