```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style='whitegrid')

date_columns = ['last_scraped', 'calendar_last_scraped', 'first_review', 'last_review']
listings = pd.read_csv('data/listings_clean.csv', parse_dates=date_columns)

print(listings.shape)
```

```python
price_night = listings['price']
no_price = listings['price'].isna()
long_min = listings['minimum_nights'] > 7
price_long = listings.loc[long_min, 'price']
vlong_min = listings['minimum_nights'] > 28
price_vlong = listings.loc[vlong_min, 'price']
percent_price = (price_night.count() / listings['id'].count()).round(2) * 100
cases = [('base', price_night), 
         ('long', price_long), 
         ('very long', price_vlong)]

print('listings with price:', price_night.count(),
      'percent of all:', percent_price)

for desc,case in cases:
    print(f'{desc} mean:', case.mean().round(1),
          f'{desc} median:', case.median())
```
