### Ex 2

```python
loans['genre'].value_counts()
```

```python
loans['genre'].nunique()
```

```python
clean_genre = (loans['genre'].str.strip().str.lower().str.title()
    .str.replace('Sci Fi', 'Science Fiction')
    .str.replace('Sci-Fi', 'Science Fiction')
    .str.replace('Children\'S', 'Children')
)
clean_genre.unique()
```

```python
loans['genre'] = clean_genre

loans['genre'].value_counts()
```

### Ex 3

```python
print(loans['returned'].isna().sum())
loans['returned'].unique()
```

```python
loans['returned'].value_counts()
```

```python
clean_returned = (loans['returned'].str.strip().str.lower()
    .str.replace('y', 'yes')
    .str.replace('n', 'no')
    .str.replace('yeses', 'yes')
    .str.replace('noo', 'no')
)

clean_returned.value_counts()
```

```python
bool_returned = clean_returned.map({'yes': True, 'no': False})

print(bool_returned.isna().sum())
bool_returned.value_counts()
```

```python
loans['returned'] = bool_returned

loans['returned'].value_counts()
```

#### Everything accounted for and converted.

### Ex 4

#### loans['late_fee'].astype(float) can not convert str

```python
cleaned_fee = (loans['late_fee'].str.strip()
    .str.replace('kr', '')
    .str.replace(' ', '')
    .str.replace(',', '.')
)

cleaned_fee.value_counts()
```

```python
float_fee = cleaned_fee.astype(float)

loans['late_fee'] = float_fee

print(loans['late_fee'].isna().sum())
loans['late_fee'].describe().round(2)
```

```python
loans['late_fee'].sum()
```

#### dtype: float64. sum of late_fee: 7677.5

```python
loans[loans.duplicated(keep=False)].head(6)
fee_with_dupli = loans['late_fee'].sum()
```
