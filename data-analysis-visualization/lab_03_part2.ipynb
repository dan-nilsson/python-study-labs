```python
cleaned_return_date = pd.to_datetime(loans['return_date'], errors='coerce')

print(cleaned_return_date.isna().sum())
cleaned_return_date.value_counts()
```

#### 30 loans have no return date. Not missing data. Just not returned.

```python
cleaned_loan_date = pd.to_datetime(loans['loan_date'], errors='coerce')

print(cleaned_loan_date.isna().sum())
cleaned_loan_date.value_counts()
```

```python
loan_delta = (cleaned_return_date - cleaned_loan_date).dt.days
minus_loan_delta = loan_delta < 0

loans.loc[minus_loan_delta, ['loan_date','return_date']]
```

#### 6 loans have negative loan period delta

```python
loans['return_date'] = cleaned_return_date
loans['loan_date'] = cleaned_loan_date
loans = loans[~minus_loan_delta]

loans.shape
```

#### merged all cleaned data. removed the 6 faulty date loans
