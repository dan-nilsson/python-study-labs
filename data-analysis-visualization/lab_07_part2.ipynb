```python
import pandas as pd
import matplotlib.pyplot as plt 
import seaborn as sns

sns.set_theme(style='whitegrid')

date_columns = ['last_scraped', 'calendar_last_scraped', 'first_review', 'last_review']
listings = pd.read_csv('data/listings_clean.csv', parse_dates=date_columns)

print(listings.shape)
```

## Lab 7 - Part 2
### Ex 1

```python
rated = listings[listings['has_reviews']]

fig, ax = plt.subplots(figsize=(9, 5))
sns.scatterplot(data=rated, x='number_of_reviews', y='review_scores_rating', alpha=0.3, ax=ax)

ax.set_xlim(0, 300) 
ax.set_ylim(0.9, 5.1)

ax.set_title('Review scores more diverse the lower the number of reviews.'+
            '\nHigher number of reviews diverge to range 4.5 - 5.0.'+
            '\n(excludes 90 review scores for listings with more than 300 reviews)')
ax.set_xlabel('Number of reviews')
ax.set_ylabel('Review score rating')

plt.show()
```

#### Funnel shaped data. Under 50 reviews it has some diversity. Above that reviews under 4.5 are almost gone. At the top-end diverges towards 5.0.
### Ex 2

```python
listings.groupby('host_is_superhost')['review_scores_rating'].agg(['median', 'mean', 'count'])

fig, ax = plt.subplots(figsize=(8, 5))

sns.boxplot(data=listings, x='host_is_superhost', y='number_of_reviews', ax=ax)
ax.set_xticks([0, 1]) 
ax.set_xticklabels(['Other hosts', 'Superhosts'])

ax.set_title('There are no super hosts with less than 300 ratings.'+
            '\nOtherwise the distrubution is very similar.')
ax.set_xlabel('')
ax.set_ylabel('Number of reviews')

plt.tight_layout()
plt.show()
```

#### Looking at the distrubution super hosts look like ordinary hosts with a requirement of 300 reviews.
