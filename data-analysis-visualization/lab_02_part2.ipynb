```python
import pandas as pd
import numpy as np

listings = pd.read_csv('data/listings.csv')
```

## Lab 2 - Part 2
### Ex 1

```python
listings['has_reviews'] = np.where(listings['number_of_reviews'] == 0, 'True', 'False')
listings['no_reviews'] = listings['number_of_reviews'] == 0

listings['has_reviews'].value_counts(normalize=True)
```

```python
listings['has_reviews'].value_counts()
```

#### 644 listings, 14.07% have no review

```python
answer = (
    listings.groupby('neighbourhood_cleansed')
    .agg(n_listings=('id', 'count'), no_reviews=('no_reviews', 'mean'))
    .sort_values('no_reviews', ascending=False)
    .round(3)
)

answer
```

#### Lowest share of no reviews is Södermals with 7.6%. Highest is Hässelby-Vällingby with 27%.

### Ex 2

```python
listings['reviewed'] = listings['number_of_reviews'] >= 1

group_reviews = pd.qcut(listings['number_of_reviews'][listings['reviewed']], 4, labels=['few', 'some', 'many', 'most'])
group_reviews.value_counts()
```

```python
pd.qcut(listings['number_of_reviews'][listings['reviewed']], 4).cat.categories
```

#### They are not equal, because many listings have exactly the same number of reviews and ties cannot be split
#### Borders: 1, 4, 13, 36, 1293

### Ex 4

```python
group_zone_type = listings.groupby(['neighbourhood_cleansed', 'room_type']).agg(n=('id', 'count'),avg=('accommodates','mean'))

group_zone_type.sort_values('avg',ascending=False).head(3)
```

```python
group_zone_type_30plus = group_zone_type['n'] >= 30

group_zone_type[group_zone_type_30plus].sort_values('avg',ascending=False).head(3)
```

#### there are 35 combinations. Without the filter the top is Östermalms / Shared room (9.00 guests, but only 4 listings). With at least 30 listings 27 combinations are left, and the top is Spånga-Tensta / Entire home/apt (5.33 guests, 54 listings).
