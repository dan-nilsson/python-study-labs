```python
import pandas as pd
```

```python
listings = pd.read_csv('data/listings.csv')
```

```python
central = ['Södermalms', 'Östermalms', 'Norrmalms', 'Kungsholmens']
```

### Ex 1

```python
mask = listings['neighbourhood_cleansed'].isin(central)
mask.mean()
```

### share of listings in central location = 50.961%

```python
pd.set_option('display.precision', 2)
central_private = listings[mask]['room_type'] == 'Private room'
not_central_private = listings[~mask]['room_type'] == 'Private room'
print(central_private.mean()*100, '%')
print(not_central_private.mean()*100, '%')
```

### Higher outside the city centre

### Ex 2

```python
entire_home = listings['room_type'] == 'Entire home/apt'
accom_4 = listings['accommodates'] >= 4
```

```python
print(accom_4[entire_home].mean())
print(accom_4[~entire_home].mean())
```

### Ex 3

```python
top = (
    listings[
        (listings['review_scores_rating'] == 5.0)
        & (listings['number_of_reviews'] >= 50)
    ]
    .head(5)[['name', 'neighbourhood_cleansed', 'number_of_reviews','review_scores_rating']]
)
top.nlargest(5, 'review_scores_rating')
```

```python
top.sort_values('review_scores_rating', ascending=False)
```

```python
top.nlargest(5, 'number_of_reviews')
```

```python
top.sort_values('number_of_reviews', ascending=False)
```

### Sorting on review score. Since they are all 5.0 it sorts on ID

### Ex 5

```python
no_bed = listings['bedrooms'] == 0
listings[no_bed].sort_values('room_type', ascending=False)[['name','room_type','bedrooms']]
```

### Most listings with 0 bedrooms are Shared room. So no private bedroom.

```python
no_avail = listings['availability_365'] == 0
print(no_avail[(listings['number_of_reviews'] == 0)].mean())
listings[no_avail].sort_values('number_of_reviews', ascending=False)[['name','room_type','price','number_of_reviews']]
```

### 35.9% are no availability and 0 reviews. The ones with reviews are probably just fully booked and long time bookings.

### The ones with 0 review are probably information by request. Probably people side-stepping the system and booking outside of it.
