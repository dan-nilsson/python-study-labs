```python
import pandas as pd
import numpy as np

listings = pd.read_csv('data/listings.csv')
```

## Part 1
### Ex 1

```python
listings['property_type'].value_counts(normalize=True).head().round(3)*100
```

#### 53.4% of the listings are of property type Entire rental unit.

### Ex 2

```python
listings['neighbourhood_cleansed'].value_counts().round(3)
```

#### Zones with the least amount of listings are:
1. Språnga-Tensta : 89
2. Skärholmens : 96
3. Rinkeby-Tensta : 108

### Ex 3

```python
listings.groupby('room_type')['beds'].agg(['count', 'mean']).round(2)
```

#### Shared room has very low count. So the avg can be very volatile. Hotel room is very low count but even 1.0 so we can assume that it's exact.

### Ex 4

```python
(listings.groupby('neighbourhood_cleansed')
    ['number_of_reviews']
    .agg(['count','mean','median','max'])
    .sort_values('median', ascending=False)
    .round(2)
)
```

#### Probably uneven distrubution. Top heavy listings with many reviews and many listings without reviews.

### Ex 5

```python
summary = listings.groupby('room_type').agg(
    n_listings=('id', 'count'),
    avg_beds=('beds', 'mean'),
    max_reviews=('number_of_reviews', 'max')
).round(2)

summary
```

### Ex 6

```python
(pd.crosstab(listings['neighbourhood_cleansed'], 
             listings['host_is_superhost'], 
             normalize='index')*100).sort_values('t',ascending=False).round(1)
```

#### Södermals has the highest percentage of super hosts with 23.8%. Brommas has the lowest at 7.1%.

### Ex 7a

```python
listings['has_reviews'] = np.where(listings['number_of_reviews'] > 0, 'True', 'False')

listings['has_reviews'].value_counts()
```

### Ex 7b

```python
listings['popularity'] = np.where(listings['number_of_reviews'] >= 50, 'popular', 'other')

listings['popularity'].value_counts()
```
